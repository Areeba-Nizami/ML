```python
#Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
'''Exploratory Data Analysis (EDA):
o	Data Type Inspection: Verify and display the data types of all features
within the loaded dataset.'''
print("---DATASET INFORMATION")
df.info()

'''o	Summary Statistics: Generate and present descriptive statistics (e.g., mean,
median, standard deviation, quartiles) for numerical columns.'''

print("\n----MEAN OF NUMERIC COLUMNS")
print(df.mean(numeric_only=True))
print("\n---MEDIAN OF NUMERIC COLUMNS")
print(df.median(numeric_only=True))
print("\n---STANDARD DEVIATION OF NUMERIC COLUMNS")
print(df.std(numeric_only=True))
print("\n---QUANTILES OF NUMERIC COLUMNS")
print(df.quantile([0.25,0.5,0.75],numeric_only=True))


'''o	Missing Value Assessment: Identify and quantify missing values for
each column, providing a clear summary of their presence.'''

print("\n---MISSING VALUES")
print(df.isna().sum())

'''o	Unique Value Counts: Determine and display the number of unique entries
for all relevant categorical and discrete numerical features. '''

print("\n---UNIQUE VALUES")
print(df.nunique())
```

```python
'''3.	Data Visualization:
o	Categorical Variable Visualization: Create bar plots for all categorical
variables to visualize their distributions and frequencies.'''

print("---BAR PLOT Pclass")
counts=df['Pclass'].value_counts()
plt.bar(counts.index,counts.values,width=0.4)
plt.title('PCLASS')
plt.show()

print("--- BAR PLOT Sex")
counts=df['Sex'].value_counts()
plt.bar(counts.index,counts.values,width=0.4)
plt.title('SEX')
plt.show()

print("---BAR PLOT Survived")
counts=df['Survived'].value_counts()
plt.bar(counts.index,counts.values,width=0.4)
plt.title('Survived')
plt.show()

print("---BAR PLOT Embarked")
counts=df['Embarked'].value_counts()
plt.bar(counts.index,counts.values,width=0.4)
plt.title('Embarked')
plt.show()

'''o	Numerical Variable Visualization: Generate scatter plots or histograms for
 numerical variables to illustrate their distributions, relationships, or trends. '''
print("---Histogram: Age")
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.hist(df['Age'].dropna())
plt.title('Distribution of Age')

print("---Histogram: Fare")
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.hist(df['Fare'].dropna())
plt.title('Distribution of Fare')
plt.show()

'''o	Correlation Heatmap: Construct a correlation heatmap for all numerical
 features to display the pairwise correlation coefficients. '''
numeric_df=df.select_dtypes(include=['int64', 'float64'])
corr_matrix=numeric_df.corr()
sns.heatmap(corr_matrix,annot=True)
plt.show()
```
